## <span style="color:#57a52d">■</span> O que é o Manim?

* *Mathematical Animation Engine*: biblioteca Python para animações matemáticas
* Criada em 2015 por Grant Sanderson para o canal [3Blue1Brown](https://www.youtube.com/c/3blue1brown) (veja a série *Essence of Calculus*)
* Animações são **programadas**: cada objeto, posição e movimento é descrito em código

* Há duas versões principais:
  - **ManimCE** (*Community Edition*): mantida pela comunidade, bem documentada: **é a que usaremos**
  - **ManimGL**: versão pessoal do autor, baseada em OpenGL

## <span style="color:#57a52d">■</span> Por que animar o Cálculo?

* O Cálculo trata de **variação**: limite, derivada e integral são processos, não figuras paradas
* No quadro, "$h \to 0$" ou "$n \to \infty$" ficam estáticos; na animação, o aluno **vê** a secante virar tangente e os retângulos preencherem a área
* A animação mostra o raciocínio passo a passo, no ritmo do professor
* O código pode ser reaproveitado: trocando a função, trocamos a animação inteira

Ao longo do minicurso, vamos animar: **derivada**, **soma de Riemann**, e **outras animações**.

## <span style="color:#57a52d">■</span> Olá, Cálculo!

A mágica `%%manim` renderiza a cena definida na célula e exibe o vídeo logo abaixo:

````python
%%manim -qm -v WARNING OlaCalculo
# A 1ª linha é a "mágica" do Jupyter. Ela PRECISA ser a primeira linha da célula:
#   %%manim            -> diz ao Jupyter: "o código desta célula é uma cena do Manim, renderize e mostre o vídeo"
#   -qm                -> qualidade média (720p, 30 quadros/s)
#   -v WARNING         -> mostra só avisos, sem as mensagens de progresso
#   OlaCalculo         -> NOME DA CLASSE que será renderizada (tem de ser igual ao nome da classe abaixo)

class OlaCalculo(Scene):                  # toda cena é uma classe que herda de Scene (é a "tela" onde tudo acontece)
    def construct(self):                  # construct é o roteiro: o Manim o executa, de cima para baixo, ao renderizar
        texto = Text("Olá, Cálculo!")     # cria um objeto de texto (ainda não aparece: só foi criado)
        self.play(Write(texto))           # play executa uma animação; Write "escreve" o texto na tela
        self.wait(1.5)                    # pausa 1,5 s para o público ler
        self.play(FadeOut(texto))         # faz o texto desaparecer gradualmente
````

## <span style="color:#57a52d">■</span> Por que `%%manim` na primeira linha?

* `%%` indica uma **mágica de célula** do Jupyter: ela atua sobre a célula **inteira**, por isso deve ser a primeira linha
* O `%%manim` faz o trabalho que, no terminal, seria feito por `manim -pqm arquivo.py OlaCalculo`:
  1. executa o código da célula
  2. renderiza a cena escolhida (a classe cujo nome vem no fim da linha)
  3. exibe o vídeo abaixo da célula
* Sem ele, a célula apenas **define** a classe: nada é desenhado

* O nome no fim da linha deve coincidir com o da classe; se errar, aparece `KeyError`
* Uma célula pode ter várias classes, mas só renderiza a que foi nomeada

## <span style="color:#57a52d">■</span> Anatomia de uma cena

* `class OlaCalculo(Scene)`: uma cena é uma **classe** derivada de `Scene`
* `construct`: método executado na renderização; é o "roteiro" da cena
* `Text("Olá, Cálculo!")`: cria um objeto visual (*Mobject*)
* `self.play(Write(texto))`: executa uma animação (*Animation*)
* `self.wait(1.5)`: pausa por 1,5 segundo

Em Python, o que significa `(Scene)` após o nome da classe? E por que todo método começa com `self`?

## <span style="color:#57a52d">■</span> Por que `construct` e por que `self`?

* **`construct`**: o Manim não sabe o que você quer animar. Ao renderizar, ele cria a sua cena e chama `construct()`; por isso o nome é **fixo**. Se você o chamar de outro nome, a cena sai vazia
* **`self`**: é a **própria cena**. Como a cena guarda o vídeo, o tempo e os objetos que estão na tela, precisamos dela para dar ordens:
  - `self.play(...)` anima e registra a animação no vídeo
  - `self.add(...)` coloca objetos na tela, sem animar
  - `self.wait(...)` deixa o tempo passar
* Criar um objeto (`texto = Text(...)`) **não** o mostra: ele só aparece quando entra em um `self.play` ou em um `self.add`

## <span style="color:#57a52d">■</span> Herança

* Uma classe pode ser **derivada** de outra, herdando seus atributos e métodos
* Exemplo com Cálculo: uma classe `Funcao`, com derivada numérica

````python
class Funcao:
    """Uma função real de uma variável."""

    def __init__(self, f, nome="f"):
        self.f = f                 # guarda a função Python (por exemplo, uma lambda)
        self.nome = nome           # nome para exibir

    def __call__(self, x):         # __call__ deixa chamar o objeto como se fosse função: g(2)
        return self.f(x)

    def derivada(self, x, h=1e-6):
        return (self(x + h) - self(x - h)) / (2*h)   # razão incremental simétrica
````

## <span style="color:#57a52d">■</span> Herança

Um polinômio **é uma** `Funcao`, com algo a mais (os coeficientes):

````python
class Polinomio(Funcao):                 # (Funcao): Polinomio herda tudo o que Funcao tem
    """Polinômio dado pelos coeficientes [c0, c1, c2, ...]."""

    def __init__(self, coefs, nome="p"):
        self.coefs = coefs
        # super() chama o __init__ da classe-mãe (Funcao), que guarda f e nome
        super().__init__(lambda x: sum(c * x**k for k, c in enumerate(coefs)), nome)

    def __repr__(self):                  # substitui o __repr__ padrão
        return f"{self.nome}: coeficientes {self.coefs}"
````

````python
p = Polinomio([0, -1, 0, 1/3], "p")      # p(x) = -x + x³/3
p, p(3), p.derivada(2), isinstance(p, Funcao)   # valor, derivada e teste de herança
````

`__call__` e `derivada` não foram escritos em `Polinomio`: foram **herdados** de `Funcao`. ($p(3) = 6$ e $p'(2) = 3$.)

## <span style="color:#57a52d">■</span> Demonstrações: onde vamos chegar

Quatro cenas só para **ver** o que o Manim faz. Não se preocupe com o código agora: objetos e posicionamento são a Parte 2, animações a Parte 3, e gráficos com `Axes` e `ValueTracker` ficam para o **Dia 2**.

````python
%%manim -qm -v WARNING Vinheta
class Vinheta(Scene):
    def construct(self):
        cores = [SEPEC_VERDE, SEPEC_AZUL, SEPEC_LARANJA, SEPEC_AMARELO]
        n = 32                                                         # número de "fios" do anel
        fios = VGroup(*[
            Line(1.15*RIGHT, 1.75*RIGHT, stroke_width=10, color=cores[k * 4 // n])   # 8 fios de cada cor
            .rotate(TAU * k / n, about_point=ORIGIN)                   # gira cada fio em torno da origem
            for k in range(n)
        ]).shift(1.3*UP)
        titulo = Text("Animando o Cálculo com o", font_size=40, weight=BOLD).next_to(fios, DOWN, buff=0.6)
        manim = Text("Manim", font_size=40, weight=BOLD)
        caixa = SurroundingRectangle(manim, color=SEPEC_LARANJA, fill_color=SEPEC_LARANJA, fill_opacity=1, buff=0.12, corner_radius=0.1)
        marca = VGroup(caixa, manim).next_to(titulo, DOWN, buff=0.25)
        faixa = VGroup(*[Rectangle(width=1.5, height=0.08, color=c, fill_color=c, fill_opacity=1, stroke_width=0) for c in cores])
        faixa.arrange(RIGHT, buff=0).next_to(marca, DOWN, buff=0.5)
        rodape = Text("III SEPEC · UFCA", font_size=24, color=SEPEC_AMARELO).next_to(faixa, DOWN, buff=0.3)

        self.play(LaggedStart(*[GrowFromPoint(f, ORIGIN + 1.3*UP) for f in fios], lag_ratio=0.05), run_time=2)  # os fios "tecem" o anel
        self.play(Rotate(fios, angle=-PI/2, about_point=fios.get_center()), Write(titulo))
        self.play(FadeIn(caixa), Write(manim))
        self.play(LaggedStart(*[GrowFromEdge(r, LEFT) for r in faixa], lag_ratio=0.8), FadeIn(rodape, shift=UP))
        self.wait(1.5)
````

````python
%%manim -qm -v WARNING ManimCELogo
# Exemplo da galeria oficial: docs.manim.community (Example Gallery)
# Sem self.play, a cena não tem animação: o Manim salva uma imagem (PNG) em vez de vídeo
class ManimCELogo(Scene):
    def construct(self):
        self.camera.background_color = "#ece6e2"
        logo_green = "#87c2a5"
        logo_blue = "#525893"
        logo_red = "#e07a5f"
        logo_black = "#343434"
        ds_m = MathTex(r"\mathbb{M}", fill_color=logo_black).scale(7)
        ds_m.shift(2.25 * LEFT + 1.5 * UP)
        circle = Circle(color=logo_green, fill_opacity=1).shift(LEFT)
        square = Square(color=logo_blue, fill_opacity=1).shift(UP)
        triangle = Triangle(color=logo_red, fill_opacity=1).shift(RIGHT)
        logo = VGroup(triangle, square, circle, ds_m)  # a ordem importa: quem vem depois fica por cima
        logo.move_to(ORIGIN)
        self.add(logo)
````

````python
%%manim -qm -v WARNING DefinicaoDerivada
# Mesma estrutura: %%manim na 1ª linha + nome da classe que será renderizada.

class DefinicaoDerivada(Scene):
    def construct(self):
        # Text: texto comum. font_size controla o tamanho; to_edge(UP) cola o objeto na borda superior da tela
        titulo = Text("Derivada", font_size=48).to_edge(UP)

        # MathTex: fórmula em LaTeX. O "r" antes das aspas (raw string) impede o Python de interpretar as barras \
        definicao = MathTex(r"f'(x) = \lim_{h \to 0} \frac{f(x+h) - f(x)}{h}")

        # next_to(objeto, direção, buff=folga): posiciona ao lado de outro objeto (aqui, abaixo da fórmula) em vez de usar coordenadas fixas
        legenda = Text("inclinação da reta tangente", font_size=30, color=YELLOW).next_to(definicao, DOWN, buff=0.8)

        self.play(Write(titulo))          # Write: ideal para textos e fórmulas
        self.play(Write(definicao))
        self.play(FadeIn(legenda))        # FadeIn: surge gradualmente
        self.wait(2)                      # tempo para ler antes de terminar o vídeo
````

Na `DefinicaoDerivada`:
* `next_to(definicao, DOWN, buff=0.8)` e `to_edge(UP)` posicionam os objetos (veja a Parte 2)
* `color=YELLOW` e `font_size=30` são **atributos** definidos na criação

````python
%%manim -qm -v WARNING Derivada
# PRÉVIA DO DIA 2: Axes, ValueTracker e always_redraw serão explicados na Parte 4

class Derivada(Scene):
    def construct(self):
        # Cria os eixos cartesianos
        eixos = Axes(
            x_range=[0, 10, 1],
            y_range=[0, 20, 5],
            tips=False,
            axis_config={
                "include_numbers": True
            }
        )

        # Identifica os eixos
        rotulo_x = eixos.get_x_axis_label("x")
        rotulo_y = eixos.get_y_axis_label("y")

        # Define a função matemática
        # (cuidado: em Python, (x/4)**1/2 é ((x/4)**1)/2, e não a raiz quadrada; por isso escrevemos (x/4)/2)
        def funcao(x):
            return -((np.sin(x - 4) * (x / 4) / 2 * (x + 3)) + (0.1 * (x - 2)**2 * np.sin(x) + 7) - 15)

        # Constrói o gráfico da função
        grafico = eixos.plot(
            funcao,
            x_range=[0, 10],
            color=BLUE
        )

        # Adiciona os eixos à cena
        self.add(eixos)

        # Desenha o gráfico
        self.play(
            Create(grafico),
            run_time=1.5
        )
        eixos.add(grafico)
        self.wait(2)

        # Controla a posição inicial e o deslocamento
        x = ValueTracker(7.3)
        dx = ValueTracker(2)

        # Cria o primeiro ponto sobre a curva
        ponto1 = always_redraw(
            lambda: Dot(color=GREEN)
            .scale(.8)
            .move_to(eixos.c2p(
                x.get_value(),
                grafico.underlying_function(x.get_value())
            ))
        )

        # Cria as projeções do primeiro ponto
        linhas_ponto1 = always_redraw(
            lambda: eixos.get_lines_to_point(
                ponto1.get_center(),
                color=GREEN
            )
        )

        # Cria o segundo ponto sobre a curva
        ponto2 = always_redraw(
            lambda: Dot(color=ORANGE)
            .scale(.8)
            .move_to(eixos.c2p(
                x.get_value() + dx.get_value(),
                grafico.underlying_function(x.get_value() + dx.get_value())
            ))
        )

        # Cria as projeções do segundo ponto
        linhas_ponto2 = always_redraw(
            lambda: eixos.get_lines_to_point(
                ponto2.get_center(),
                color=ORANGE
            )
        )

        # Constrói a reta secante e os incrementos
        secante = always_redraw(
            lambda: eixos.get_secant_slope_group(
                x=x.get_value(),
                dx=dx.get_value(),
                graph=grafico,
                dx_label=r"\Delta{x}",
                dy_label=r"\Delta{y}",
                dx_line_color=GREEN,
                dy_line_color=ORANGE,
                secant_line_color=YELLOW,
                secant_line_length=5
            )
        )

        # Exibe o primeiro ponto e suas projeções
        self.play(
            Create(ponto1),
            Create(linhas_ponto1),
        )
        eixos.add(ponto1, linhas_ponto1)
        self.wait(2)

        # Exibe o segundo ponto e suas projeções
        self.play(
            Create(ponto2),
            Create(linhas_ponto2),
        )
        eixos.add(ponto2, linhas_ponto2)
        self.wait(2)

        # Exibe a reta secante
        self.play(
            Write(secante),
            run_time=1.2
        )
        self.wait(2)

        # Remove as linhas auxiliares
        self.play(
            Uncreate(linhas_ponto1),
            Uncreate(linhas_ponto2),
        )
        self.wait()

        # Reduz o incremento horizontal
        self.play(
            dx.animate.set_value(1.5),
            run_time=5
        )
        self.wait()

        # Aproxima o incremento de zero
        self.play(
            dx.animate.set_value(1e-9),
            run_time=3
        )
        self.wait()

        # Move o ponto inicial para x = 1,5
        self.play(
            x.animate.set_value(1.5),
            run_time=4
        )
        self.wait()

        # Move o ponto inicial para x = 8
        self.play(
            x.animate.set_value(8),
            run_time=7
        )
        self.wait()

        # Remove a secante e os dois pontos
        self.play(
            Uncreate(secante),
            Uncreate(ponto1),
            Uncreate(ponto2),
        )

        # Remove o gráfico da função
        self.play(
            Uncreate(grafico),
        )
        self.wait()
````

* Formas geométricas, eixos, gráficos, textos, fórmulas, grupos de objetos, ...
* Nesta parte, usaremos `self.add`, que apenas **coloca** objetos na tela (sem animar), e a opção `-s`, que gera só uma **imagem** do último quadro: é rápido e ótimo para conferir posições

## <span style="color:#146dcc">■</span> Sistema de coordenadas

* A tela mede cerca de $14{,}2 \times 8$ unidades, com a origem no centro
* Direções pré-definidas: `UP`, `DOWN`, `LEFT`, `RIGHT` (vetores unitários), `UL`, `UR`, `DL`, `DR` e `ORIGIN`
* Pontos são vetores $(x, y, z)$: `2*RIGHT + UP` é o ponto $(2, 1, 0)$
* O Manim é 3D por construção; em cenas 2D, o $z$ fica sempre $0$

````python
%%manim -qm -s -v WARNING Coordenadas
class Coordenadas(Scene):
    def construct(self):
        plano = NumberPlane()                      # plano cartesiano de fundo, com a malha de 1 em 1 unidade
        origem = Dot(ORIGIN, color=YELLOW)         # Dot: ponto; ORIGIN = (0, 0, 0)
        P = 2*RIGHT + UP                           # vetores se somam e se multiplicam por escalar: P = (2, 1, 0)
        p = Dot(P, color=RED)                      # um ponto colocado em P
        vetor = Arrow(ORIGIN, P, buff=0, color=RED)  # Arrow(início, fim); buff=0 evita a folga nas extremidades
        rotulo = MathTex("(2, 1)").next_to(p, UR, buff=0.1)  # rótulo no canto superior direito (UR) do ponto, com folga 0,1
        self.add(plano, origem, vetor, p, rotulo)  # add coloca tudo na tela de uma vez (a ordem define quem fica por cima)
````

## <span style="color:#146dcc">■</span> Formas geométricas

````python
%%manim -qm -s -v WARNING Formas
class Formas(Scene):
    def construct(self):
        ret = Rectangle(width=1.5, height=2.5, color=BLUE)         # retângulo (as "barras" das somas de Riemann)
        reta = Line(LEFT, RIGHT, color=YELLOW)                      # segmento de reta entre dois pontos
        seta = Arrow(LEFT, RIGHT, color=RED)                        # segmento com ponta (vetores)
        ponto = Dot(color=GREEN)                                    # ponto (por padrão, na origem)
        arco = Arc(radius=1, angle=PI/2, color=ORANGE)              # arco de circunferência; ângulo em RADIANOS (PI/2 = 90°)
        trap = Polygon([-1, -1, 0], [1, -1, 0], [0.5, 1, 0], [-0.5, 1, 0], color=PURPLE)  # polígono dados os vértices (trapézio)
        formas = VGroup(ret, reta, seta, ponto, arco, trap)         # VGroup agrupa objetos para tratá-los como um só
        formas.arrange(RIGHT, buff=0.6)                             # arrange enfileira o grupo lado a lado, com espaço 0,6
        self.add(formas)
````

## <span style="color:#146dcc">■</span> Posicionamento

| método | efeito |
|---|---|
| `shift(v)` | desloca pelo vetor `v` |
| `move_to(p)` | coloca o centro no ponto `p` |
| `next_to(obj, direção)` | ao lado de outro objeto |
| `to_edge(direção)` | junto a uma borda da tela |
| `to_corner(direção)` | junto a um canto da tela |
| `align_to(obj, direção)` | alinha uma borda com a de outro objeto |

````python
%%manim -qm -v WARNING Posicionamento
class Posicionamento(Scene):
    def construct(self):
        r = Rectangle(width=2, height=1.5)                               # um retângulo no centro
        altura = MathTex(r"f(x_i)").next_to(r, RIGHT, buff=0.3)          # ao lado direito do retângulo (altura)
        base = MathTex(r"\Delta x").next_to(r, DOWN, buff=0.2)           # abaixo do retângulo (base)
        titulo = Text("Posicionamento").to_edge(UP)                      # colado na borda de cima
        canto = Dot(color=YELLOW).to_corner(DL)                          # no canto inferior esquerdo
        solto = Dot(color=RED).move_to(3*LEFT + UP)                      # em um ponto exato: (-3, 1, 0)
        deslocado = Dot(color=GREEN).shift(2*UP + 3*RIGHT)               # deslocado 2 acima e 3 à direita da posição atual (origem)

        # Animando cada elemento em sequência para gerar o vídeo
        self.play(Write(titulo))
        self.play(Create(r))
        self.play(Write(altura), Write(base))
        self.play(FadeIn(canto, scale=1.5), FadeIn(solto, scale=1.5), FadeIn(deslocado, scale=1.5))
        self.wait(1.5)
````

* Esses métodos retornam o próprio objeto, por isso podem ser **encadeados**: `MathTex("x").scale(0.5).next_to(r, UP)`
* Prefira posições **relativas** (`next_to`, `align_to`): se o objeto mudar de tamanho, o resto acompanha

## <span style="color:#146dcc">■</span> Textos e fórmulas

* `Text`: texto comum (acentos funcionam, sem LaTeX)
* `MathTex`: fórmula em modo matemático do LaTeX
* `Tex`: texto LaTeX, com fórmulas entre `$...$`

````python
%%manim -qm -s -v WARNING Textos
class Textos(Scene):
    def construct(self):
        t = Text("Teorema Fundamental do Cálculo", color=YELLOW, font_size=44)   # texto comum
        f = MathTex(r"\int_a^b f'(x)\,dx = f(b) - f(a)", font_size=72)         # fórmula: r"..." evita que \f, \t, ... virem caracteres especiais
        frase = Tex(r"Se $f'$ é contínua em $[a, b]$", font_size=40)             # texto LaTeX com fórmulas entre $ $
        grupo = VGroup(t, f, frase).arrange(DOWN, buff=0.6)                       # empilha os três, de cima para baixo
        self.add(grupo)
````

* O `r` antes das aspas (*raw string*) evita que o Python interprete as barras: sem ele, o `\f` de `\frac` viraria um caractere especial
* `\,` coloca um pequeno espaço antes do `dx`; `\lim`, `\int`, `\sum`, `\frac` são comandos LaTeX comuns no Cálculo

## <span style="color:#146dcc">■</span> Partes de uma fórmula

Uma `MathTex` com vários argumentos é uma lista de partes, que podem ser acessadas por índice:

## <span style="color:#146dcc">■</span> Exemplo: o triângulo da inclinação

A inclinação de uma reta é a razão $m = \dfrac{\Delta y}{\Delta x}$; é a ideia que, no limite, dará a derivada. Vamos montar a figura **só com o que vimos**:

````python
%%manim -qm -s -v WARNING TrianguloInclinacao
class TrianguloInclinacao(Scene):
    def construct(self):
        # np.array cria os vértices como vetores (x, y, z); o np (NumPy) já vem com o "from manim import *"
        A = np.array([-2, -1, 0])
        B = np.array([2, -1, 0])
        C = np.array([2, 2, 0])

        # malha mais discreta: background_line_style muda o estilo das linhas de fundo do plano
        plano = NumberPlane(background_line_style={"stroke_opacity": 0.3})
        secante = Line(A - (C - A)*0.3, C + (C - A)*0.3, color=YELLOW)   # reta por A e C, um pouco prolongada nas pontas
        horizontal = DashedLine(A, B, color=WHITE)                       # cateto horizontal: Δx
        vertical = DashedLine(B, C, color=WHITE)                         # cateto vertical: Δy
        pontos = VGroup(Dot(A, color=RED), Dot(C, color=RED))            # os dois pontos da reta

        dx = MathTex(r"\Delta x = 4").next_to(horizontal, DOWN)          # rótulos posicionados relativamente aos catetos
        dy = MathTex(r"\Delta y = 3").next_to(vertical, RIGHT)
        m = MathTex(r"m = \frac{\Delta y}{\Delta x} = \frac{3}{4}").to_corner(UL)  # fórmula no canto da tela
        self.add(plano, secante, horizontal, vertical, pontos, dx, dy, m)
````

* Os rótulos usam `next_to`, então acompanham a figura se mudarmos `A`, `B` ou `C` (só precisamos corrigir os números do texto)
* Na Parte 4, o ponto $C$ vai se **aproximar** de $A$: a secante virará tangente

## <span style="color:#f06418">■</span> Animações

* Uma *Animation* descreve como um objeto muda ao longo do tempo
* É executada com `self.play(...)`
* Ao final, o objeto **permanece** na tela (exceto nas animações de saída, como `FadeOut`)

| animação | efeito |
|---|---|
| `Create` | traça o contorno |
| `Write` | escreve (textos e fórmulas) |
| `DrawBorderThenFill` | traça o contorno e depois preenche |
| `FadeIn`, `FadeOut` | surge ou desaparece gradualmente |
| `Uncreate`, `Unwrite` | desfazem o `Create` e o `Write` (saída) |
| `GrowFromCenter` | cresce a partir do centro |

## <span style="color:#f06418">■</span> Entradas e saídas

````python
%%manim -qm -v WARNING Entradas
class Entradas(Scene):
    def construct(self):
        r = Rectangle(width=2, height=1.5, color=BLUE)              # contorno de um retângulo
        q = Square(color=GREEN, fill_opacity=0.5)                   # quadrado preenchido (opacity>0 é preciso para DrawBorderThenFill ter o que preencher)
        t = MathTex(r"\int f(x)\,dx")                               # fórmula
        VGroup(r, q, t).arrange(RIGHT, buff=1)                      # só posiciona; não precisa guardar o grupo, pois o efeito fica nos objetos
        self.play(Create(r))                                        # Create: traça o contorno (formas)
        self.play(DrawBorderThenFill(q))                            # contorno primeiro, preenchimento depois
        self.play(Write(t))                                         # Write: textos e fórmulas
        self.play(FadeOut(r, q, t))                                 # vários objetos no mesmo play saem ao mesmo tempo
````

## <span style="color:#f06418">■</span> `rate_func`: o ritmo da animação

O `rate_func` define **como o tempo é distribuído**: `smooth` (padrão: começa e termina devagar), `linear` (velocidade constante), `there_and_back` (vai e volta), `rush_into` (acelera), `rush_from` (freia), `double_smooth`, `wiggle`, `exponential_decay`.

````python
%%manim -qm -v WARNING TaxasDeAnimacao
class TaxasDeAnimacao(Scene):
    def construct(self):
        taxas = [linear, smooth, there_and_back, rush_into, rush_from]          # são funções: passadas SEM parênteses
        ys = [2, 1, 0, -1, -2]                                                   # uma linha para cada taxa
        pontos = [Dot([-3.5, y, 0], color=YELLOW) for y in ys]
        nomes = [Text(t.__name__, font_size=22).next_to(p, LEFT, buff=0.4)      # t.__name__: nome da função
                 for t, p in zip(taxas, pontos)]
        self.add(*pontos, *nomes)
        # Os cinco pontos fazem o MESMO deslocamento, na mesma duração, só com ritmos diferentes
        self.play(*[p.animate(rate_func=t).shift(9*RIGHT) for p, t in zip(pontos, taxas)], run_time=4)
        self.wait()
````

**Comparando** os pontos, vemos: `linear` (constante), `smooth` (começa e termina devagar), `there_and_back` (vai até o final e volta), `rush_into` (acelera até o fim) e `rush_from` (parte rápido e freia).

## <span style="color:#f06418">■</span> Compondo animações no tempo

````python
%%manim -qm -v WARNING Composicao
class Composicao(Scene):
    def construct(self):
        quadrados = VGroup(*[Square(0.9, fill_opacity=0.6) for _ in range(6)]).arrange(RIGHT, buff=0.3)
        # LaggedStart: começa cada animação um pouco depois da anterior; lag_ratio = fração da duração de espera (0 = juntas, 1 = em fila)
        self.play(LaggedStart(*[FadeIn(q, shift=UP) for q in quadrados], lag_ratio=0.3))
        # AnimationGroup: reúne animações em uma só (aqui, todas AO MESMO TEMPO porque lag_ratio=0)
        self.play(AnimationGroup(*[q.animate.set_color(RED) for q in quadrados], lag_ratio=0))
        # Succession: uma animação só COMEÇA quando a anterior TERMINA (estritamente em sequência)
        self.play(Succession(Indicate(quadrados[0]), Indicate(quadrados[-1])))
        self.wait()
        # Atalho para o LaggedStart de uma lista:  LaggedStartMap(FadeIn, quadrados, shift=UP, lag_ratio=0.2)
````

* `Succession(a, b)`: `b` só começa quando `a` termina
* `AnimationGroup(a, b, lag_ratio=0.5)`: sobreposição **parcial**
* `LaggedStart`: o mesmo efeito em **muitos** objetos; ótimo para barras de Riemann entrando uma a uma

## <span style="color:#f06418">■</span> Movimento ao longo de um caminho

````python
%%manim -qm -v WARNING Movimento
class Movimento(Scene):
    def construct(self):
        caminho = Arc(radius=2.5, start_angle=PI, angle=-PI, color=GREY)    # semicircunferência da esquerda para a direita, por cima
        caminho.shift(DOWN)                                                   # base da curva levemente abaixo do centro
        ponto = Dot(caminho.get_start(), color=YELLOW)                        # começa no início do caminho
        self.add(caminho, ponto)
        # MoveAlongPath: o ponto percorre qualquer curva (Line, Arc, gráfico de função, ...)
        self.play(MoveAlongPath(ponto, caminho), run_time=3, rate_func=linear)   # linear: velocidade constante
        quadrado = Square(0.8, color=BLUE).shift(3*LEFT + 2*UP)
        self.play(Create(quadrado))
        # Rotate: gira em torno do ponto escolhido (about_point); aqui, em torno da origem
        self.play(Rotate(quadrado, angle=PI, about_point=ORIGIN), run_time=2)
        # Outras formas de movimento:
        #   Rotating(obj, radians=TAU, about_point=ORIGIN)   # gira continuamente
        #   obj.animate.rotate(PI/2)                         # gira 90° (equivale a Rotate)
        #   obj.add_updater(lambda m, dt: m.shift(dt*RIGHT)) # velocidade constante: 1 unidade/s (Parte 4)
        self.wait()
````

## <span style="color:#f06418">■</span> Animações simultâneas e `.animate`

* Várias animações em um mesmo `self.play` ocorrem **ao mesmo tempo**
* `run_time` controla a duração (em segundos)
* `obj.animate.metodo(...)` anima **qualquer método** que altere o objeto (`shift`, `scale`, `set_color`, ...)

````python
%%manim -qm -v WARNING Simultaneas
class Simultaneas(Scene):
    def construct(self):
        a = Rectangle(width=2, height=1, color=BLUE).shift(2*LEFT)    # dois objetos, um à esquerda...
        b = Square(color=GREEN).shift(2*RIGHT)                         # ...e outro à direita
        self.play(Create(a), Create(b), run_time=2)                    # os dois são desenhados juntos, em 2 s
        # .animate transforma a chamada do método em animação: em vez de pular para o resultado, ele "caminha" até lá
        self.play(a.animate.shift(RIGHT).set_color(RED),               # a anda e muda de cor...
                  b.animate.scale(0.5).rotate(PI/4))                   # ...enquanto b encolhe e gira (tudo simultâneo)
        self.wait()
````

## <span style="color:#f06418">■</span> Transformações

`Transform(a, b)` deforma `a` até que ele tenha a aparência de `b`:

````python
%%manim -qm -v WARNING Transformacoes
class Transformacoes(Scene):
    def construct(self):
        f = MathTex("f(x) = x^2")                  # função
        g = MathTex("f'(x) = 2x")                  # sua derivada
        self.play(Write(f))
        self.play(Transform(f, g))                 # f passa a ter a aparência de g
        self.play(f.animate.shift(2*UP))           # CUIDADO: quem está na tela é f, não g!
        self.wait()
````

* Após `Transform(f, g)`, o objeto na tela continua sendo `f` (só com a aparência de `g`)
* `ReplacementTransform(f, g)` faz o mesmo efeito, mas **substitui** `f` por `g` na tela (depois, use `g`)

## <span style="color:#f06418">■</span> Destaques

````python
%%manim -qm -v WARNING Destaques
class Destaques(Scene):
    def construct(self):
        f = MathTex(r"f'(x)", "=", r"\lim_{h \to 0}", r"\frac{f(x+h) - f(x)}{h}", font_size=72)  # em partes (Parte 2)
        self.add(f)                                 # add: já começa na tela, sem animação de entrada
        self.play(Indicate(f[2]))                   # Indicate: aumenta e colore a parte por instantes (aqui, o limite)
        self.play(Circumscribe(f[3]))               # Circumscribe: desenha um retângulo em volta (aqui, a razão incremental)
        self.play(Circumscribe(f))                  # sem índice: destaca a fórmula inteira
        self.wait()
````

## <span style="color:#f06418">■</span> Passagens algébricas

* `TransformMatchingTex(a, b)` move as partes **iguais** de `a` e `b` e faz surgir ou sumir as demais
* Por isso, as fórmulas devem ser divididas em partes (Parte 2): as partes que se repetem ficam paradas
* Exemplo: derivar $f(x) = x^2$ **pela definição**

````python
%%manim -qm -v WARNING DerivadaPelaDefinicao
class DerivadaPelaDefinicao(Scene):
    def construct(self):
        titulo = Text(r"Derivada de f(x) = x²", font_size=36).to_edge(UP)
        # lista de passos: as 3 primeiras partes se repetem; só a última (a razão) muda a cada passo
        passos = [
            MathTex(r"f'(x)", "=", r"\lim_{h \to 0}", r"\frac{(x+h)^2 - x^2}{h}"),
            MathTex(r"f'(x)", "=", r"\lim_{h \to 0}", r"\frac{x^2 + 2xh + h^2 - x^2}{h}"),   # desenvolve o quadrado
            MathTex(r"f'(x)", "=", r"\lim_{h \to 0}", r"\frac{2xh + h^2}{h}"),                # cancela x²
            MathTex(r"f'(x)", "=", r"\lim_{h \to 0}", r"(2x + h)"),                           # divide por h
            MathTex(r"f'(x)", "=", r"2x"),                                                    # faz h → 0
        ]
        for p in passos:
            p.scale(1.3)                            # aumenta todos de uma vez (escreva o loop para não repetir código)
        atual = passos[0]
        self.play(Write(titulo))
        self.play(Write(atual))                     # o primeiro passo é escrito; os demais serão transformados
        for proximo in passos[1:]:                  # passos[1:] = do segundo em diante
            self.wait(1)                            # pausa para o aluno acompanhar
            self.play(TransformMatchingTex(atual, proximo))   # move as partes iguais, troca as diferentes
            atual = proximo                         # o objeto que está na tela agora é o "proximo"
        self.play(Circumscribe(atual))              # destaca o resultado
        self.wait()
````

* `atual = proximo`: após cada transformação, o objeto da tela passa a ser o novo (lembra do alerta de `Transform`?)
* Se algum passo parecer "estranho", confira se as **partes iguais** têm exatamente o mesmo código LaTeX

## <span style="color:#f06418">■</span> Exemplo: área sob a reta $y = x$

Vamos **ver** a integral nascer como limite de somas. Para a reta $y = x$ em $[0, 4]$, o triângulo tem área $\dfrac{4^2}{2} = 8$.

Aproximamos por retângulos (extremo direito) e aumentamos $n$: a soma $S_n = 8\left(1 + \tfrac{1}{n}\right)$ tende a $8$.

````python
%%manim -qm -v WARNING AreaSobReta
class AreaSobReta(Scene):
    def construct(self):
        O = np.array([-5, -2, 0])                                  # origem do desenho (canto inferior esquerdo)
        eixo_x = Arrow(O, O + 5*RIGHT, buff=0)                     # eixos desenhados "na mão", só com setas
        eixo_y = Arrow(O, O + 4.5*UP, buff=0)
        reta = Line(O, O + 4*RIGHT + 4*UP, color=YELLOW)           # gráfico de y = x entre 0 e 4 (1 unidade = 1 unidade)
        rot = MathTex("y = x", color=YELLOW).next_to(reta.get_end(), UL)   # get_end(): ponto final da reta

        def retangulos(n):                                          # função que devolve n retângulos de largura 4/n
            w = 4/n                                                 # largura de cada retângulo
            return VGroup(*[                                        # * "desembrulha" a lista para dentro do VGroup
                Rectangle(width=w, height=(k + 1)*w, color=BLUE,    # altura = valor da reta no extremo direito: (k+1)·w
                          fill_color=BLUE, fill_opacity=0.5, stroke_width=1)
                .move_to(O + k*w*RIGHT, aligned_edge=DL)            # coloca o canto inferior esquerdo (DL) do retângulo no ponto
                for k in range(n)                                   # k = 0, 1, ..., n-1
            ])

        self.play(Create(eixo_x), Create(eixo_y), Create(reta), Write(rot))
        barras = retangulos(4)                                      # começamos com 4 retângulos
        soma = MathTex("S_4 = 10").move_to(3.5*RIGHT + 2*UP)        # soma calculada: 8(1 + 1/4) = 10 (centrada na metade direita)
        self.play(Create(barras), Write(soma))
        self.wait()

        for n, valor in [(8, "9"), (16, "8{,}5"), (64, "8{,}125")]:  # {,} é a vírgula decimal sem espaço extra em LaTeX
            novas = retangulos(n)
            nova_soma = MathTex(f"S_{{{n}}} = {valor}").move_to(soma)   # {{ }} dentro da f-string vira { } no LaTeX
            self.play(Transform(barras, novas), Transform(soma, nova_soma))   # barras e soma mudam juntas
            self.wait(0.5)

        limite = MathTex(r"\lim_{n\to\infty} S_n = 8 = \frac{4^2}{2}").next_to(soma, DOWN, buff=0.8)   # centrado sob a soma
        self.play(Write(limite))
        self.play(Circumscribe(limite, color=YELLOW))
        self.wait(2)
````

* A função `retangulos(n)` reaproveita código: mudar $n$ não exige reescrever nada
* Na Parte 4, faremos isso com **eixos de verdade**, para qualquer função

## <span style="color:#f4b718">■</span> Colinha: direções e constantes

````python
%%manim -qm -s -v WARNING Direcoes
class Direcoes(Scene):
    def construct(self):
        # Direções prontas do Manim: todas são vetores (x, y, z). Podem ser somadas e multiplicadas: 2*UP + LEFT
        direcoes = {"UP": UP, "DOWN": DOWN, "LEFT": LEFT, "RIGHT": RIGHT,    # (0,1,0) (0,-1,0) (-1,0,0) (1,0,0)
                    "UL": UL, "UR": UR, "DL": DL, "DR": DR}                   # diagonais: Up-Left, Up-Right, Down-Left, Down-Right
        for nome, vetor in direcoes.items():                                  # percorre o dicionário (nome, vetor)
            seta = Arrow(ORIGIN, 2*vetor, buff=0, color=YELLOW)               # seta de comprimento 2 nessa direção
            rotulo = Text(nome, font_size=28).next_to(seta.get_end(), vetor, buff=0.15)   # rótulo além da ponta da seta
            self.add(seta, rotulo)
        self.add(Dot(ORIGIN, color=RED))                                      # ORIGIN = (0, 0, 0): centro da tela
        # Outras constantes úteis (não aparecem aqui):
        #   IN / OUT  -> eixo z (para dentro / para fora da tela) em cenas 3D
        #   PI, TAU   -> π e 2π (ângulos SEMPRE em radianos)
        #   DEGREES   -> converte graus em radianos: 90*DEGREES == PI/2
        #   X_AXIS, Y_AXIS, Z_AXIS -> vetores unitários dos eixos
````

## <span style="color:#f4b718">■</span> Colinha: perguntando ao objeto

Todo Mobject sabe **onde está** e **qual o seu tamanho**: ótimo para ligar um objeto ao outro sem calcular coordenadas.

| método | devolve |
|---|---|
| `get_center()` | o centro |
| `get_left()`, `get_right()`, `get_top()`, `get_bottom()` | pontos no meio de cada lado |
| `get_corner(UR)` | um canto (`UL`, `UR`, `DL`, `DR`) |
| `get_start()`, `get_end()` | início e fim (retas, setas, curvas) |
| `.width`, `.height` | largura e altura |
| `copy()` | uma **cópia independente** (útil para animar "de onde estava" até onde vai) |

````python
%%manim -qm -v WARNING Utilitarios
class Utilitarios(Scene):
    def construct(self):
        r = Rectangle(width=4, height=2.5, color=BLUE, fill_opacity=0.3)       # objeto de referência
        pontos = VGroup(                                                       # um ponto em cada lugar "perguntado"
            Dot(r.get_center(), color=WHITE),                                  # centro
            Dot(r.get_left(), color=RED), Dot(r.get_right(), color=RED),       # meio do lado esquerdo e direito
            Dot(r.get_top(), color=GREEN), Dot(r.get_bottom(), color=GREEN),   # meio do lado de cima e de baixo
            Dot(r.get_corner(UR), color=YELLOW),                               # canto superior direito
        )
        medidas = Text(f"largura = {r.width:.1f}   altura = {r.height:.1f}", font_size=28).next_to(r, DOWN, buff=0.6)
        # f"..." (f-string) insere valores do Python no texto; :.1f = 1 casa decimal

        copia = r.copy().scale(0.4).set_color(ORANGE).next_to(r, RIGHT, buff=1)   # copy(): cópia que pode ser alterada sem mexer em r

        # Animando cada etapa da construção para gerar o vídeo
        self.play(Create(r))
        self.play(Create(pontos))
        self.play(Write(medidas))
        self.play(TransformFromCopy(r, copia))
        self.wait(1.5)

        # Mais transformações geométricas (todas devolvem o próprio objeto, então encadeiam):
        #   r.scale(2)                          # escala em torno do centro (scale(0.5) encolhe)
        #   r.scale_to_fit_width(3)             # ajusta a largura mantendo a proporção (scale_to_fit_height também)
        #   r.stretch(2, 0)                     # estica só no eixo x (dim=0); dim=1 estica na altura
        #   r.flip(UP)                          # espelha (inverte)
        #   r.rotate(PI/6)                      # gira 30° em torno do centro; about_point=... muda o centro de giro
        #   r.set_opacity(0.5)                  # opacidade geral (contorno e preenchimento)
        #   r.set_z_index(5)                    # camada: maior fica por cima, independentemente da ordem do add
````

## <span style="color:#f4b718">■</span> Mover: seis formas distintas

Há **vários métodos** para posicionar; cada um serve a uma situação. A cena abaixo leva seis pontos para a esquerda:

* os três primeiros usam **coordenadas** e terminam todos em $x = -3$;
* os três últimos usam **referências**: o ponto cinza, a borda da tela e a linha tracejada.

````python
%%manim -qm -v WARNING MoverParaEsquerda
class MoverParaEsquerda(Scene):
    def construct(self):
        ys = [2.5, 1.5, 0.5, -0.5, -1.5, -2.5]                        # uma "linha" para cada método
        cores = [RED, ORANGE, YELLOW, GREEN, BLUE, PURPLE]
        pontos = [Dot([0, y, 0], color=c) for y, c in zip(ys, cores)]

        alvo = Dot([-2, ys[3], 0], color=GREY)                        # referência da 4ª linha (next_to)
        guia = DashedLine([-5, 3.2, 0], [-5, -3.2, 0], color=GREY)    # referência da 6ª linha (align_to): reta x = -5

        nomes = [
            "shift(3*LEFT)",              # coordenadas: termina em x = -3
            "move_to([-3, y, 0])",
            "set_x(-3)",
            "next_to(alvo, LEFT)",        # referências: termina onde a referência mandar
            "to_edge(LEFT)",
            "align_to(guia, LEFT)",
        ]
        rotulos = [Text(n, font_size=22).move_to([3, y, 0]) for n, y in zip(nomes, ys)]

        # 1. Mostra os pontos no centro, as referências e os nomes
        self.play(FadeIn(alvo), Create(guia), *[FadeIn(p) for p in pontos], *[Write(r) for r in rotulos])
        self.wait(0.5)

        # 2. Move os seis pontos ao mesmo tempo
        self.play(
            pontos[0].animate.shift(3*LEFT),                    # anda 3 unidades para a esquerda (relativo)
            pontos[1].animate.move_to([-3, ys[1], 0]),          # vai para um ponto exato (absoluto)
            pontos[2].animate.set_x(-3),                        # muda só a coordenada x
            pontos[3].animate.next_to(alvo, LEFT),              # fica ao lado esquerdo do ponto cinza
            pontos[4].animate.to_edge(LEFT),                    # encosta na borda esquerda da tela
            pontos[5].animate.align_to(guia, LEFT),             # borda esquerda do ponto alinhada à da linha tracejada
            run_time=2
        )
        self.wait(1.5)
````

**Quando usar cada um**
* `shift`: "ande 2 para a direita" (relativo) · `move_to`: "vá para ali" (absoluto)
* `next_to`, `align_to`, `to_edge`, `to_corner`: layout que **acompanha** o objeto de referência e a tela
* `arrange`, `arrange_in_grid`: vários objetos de uma vez

## <span style="color:#f4b718">■</span> Catálogo de animações

| grupo | animações |
|---|---|
| **entrada** | `Create`, `Write`, `DrawBorderThenFill`, `FadeIn(obj, shift=UP)`, `GrowFromCenter`, `GrowFromPoint`, `GrowArrow`, `SpinInFromNothing` |
| **saída** | `FadeOut`, `Uncreate` (desfaz o `Create`), `Unwrite` (desfaz o `Write`) |
| **transformação** | `Transform`, `ReplacementTransform`, `TransformFromCopy`, `FadeTransform`, `TransformMatchingTex` |
| **movimento** | `obj.animate...`, `MoveAlongPath`, `Rotate(obj, angle, about_point)`, `Rotating` |
| **destaque** | `Indicate`, `Circumscribe`, `Flash`, `FocusOn`, `Wiggle`, `ApplyWave`, `ShowPassingFlash` |
| **composição** | `AnimationGroup`, `LaggedStart`, `Succession` |

````python
%%manim -qm -v WARNING CatalogoDeAnimacoes
class CatalogoDeAnimacoes(Scene):
    def construct(self):
        a = Square(color=BLUE, fill_opacity=0.5)
        self.play(GrowFromCenter(a))                              # cresce a partir do centro
        self.play(Rotate(a, angle=PI/2))                          # gira 90° em torno do centro (about_point=... muda o eixo de giro)
        b = Circle(color=GREEN, fill_opacity=0.5).shift(3*RIGHT)
        self.play(TransformFromCopy(a, b))                        # b nasce como CÓPIA de a se transformando; a continua na tela
        seta = Arrow(LEFT, RIGHT, color=YELLOW).shift(3*LEFT)
        self.play(GrowArrow(seta))                                # seta cresce a partir da origem
        t = Text("Calculo", font_size=36).shift(2*DOWN)
        self.play(FadeIn(t, shift=UP))                            # surge deslizando para cima (shift=direção de onde vem)
        t2 = Text("Cálculo!", font_size=36, color=YELLOW).shift(2*DOWN)
        self.play(FadeTransform(t, t2))                           # troca por outro objeto com um "fade" suave
        self.play(Wiggle(t2), Flash(b))                           # destaques: tremer e dar um "flash" no ponto/objeto
        self.play(ApplyWave(a))                                   # onda que percorre o objeto
        self.play(Uncreate(a), Unwrite(t2), FadeOut(b, seta))     # saídas: desfaz o traço, "desescreve", some
        self.wait()
````

* Cada animação aceita `run_time=` (duração), `rate_func=` (ritmo) e, nas de grupo, `lag_ratio=`
